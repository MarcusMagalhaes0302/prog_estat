```python
import numpy as np
import math
```

```python
#Binomial
def binomial(n,p):
    uniforme = np.random.uniform(size = n)
    return np.sum(uniforme<p)
```

```python
#Exponencial
def exponencial(lamb):
    u = np.random.uniform()
    return -(1/lamb)*np.log(1-u)
```

```python
#Poisson
def poisson(lam):
    u = np.random.uniform()
    lam = 3
    i = 0
    p = np.exp(-lam)
    F = p

    while u>F:
        i = i+1
        p = p*(lam/i)
        F = F + p

    x = i
    return x
```

```python
#Gama para alpha >1
def Gama(lamb,alpha):
    def g(x,mi):
        return mi*np.exp(-mi*x)

    def f(x,lamb,alpha):
        num = (lamb ** alpha)*(x**(alpha - 1))*np.exp(-lamb*x)
        den = math.gamma(alpha)
        return num/den
    mi = lamb / alpha
    xstar = (alpha - 1)/(lamb - mi)
    c = f(xstar,lamb,alpha)/g(xstar,mi)

    u = np.random.uniform()
    y = -(1/mi)*np.log(1-u)
    U = np.random.uniform()

    while c*U*g(y,mi) > f(y,lamb,alpha):
        u = np.random.uniform()
        y = -(1/mi)*np.log(1-u)
        U = np.random.uniform()

    G = y
    uni = np.random.uniform()
    X = G*((uni)**(1/(alpha-1)))
    return X


#Gama para alpha inteiro
def Gama_inteiro(k,lamb):
    gama = []
    for i in range(k):
        u = np.random.uniform()
        exp = -(1/lamb)*np.log(1-u)
        gama.append(exp)
        
    return np.sum(gama)

#Gama para alpha < 1
def Gama_menor(lamb,alpha):
    def g(x,mi):
        return mi*np.exp(-mi*x)

    def f(x,lamb,alpha):
        num = (lamb ** alpha)*(x**(alpha - 1))*np.exp(-lamb*x)
        den = math.gamma(alpha)
        return num/den

    mi = lamb / alpha
    xstar = (alpha - 1)/(lamb - mi)
    c = f(xstar,lamb,alpha)/g(xstar,mi)

    u = np.random.uniform()
    y = -(1/mi)*np.log(1-u)
    U = np.random.uniform()

    while c*U*g(y,mi) > f(y,lamb,alpha):
        u = np.random.uniform()
        y = -(1/mi)*np.log(1-u)
        U = np.random.uniform()

    G = y
    uni = np.random.uniform()
    X = G*((uni)**(1/(alpha-1)))

    return X
```

```python
#Gerando a Normal

def Normal():
    def f(x):
        numerador = np.exp((-x**2)/2)
        denominador = np.sqrt(2*np.pi)
        return numerador/denominador

    def exponencial(x):
        return np.exp(-x)

    x_estrela = 1
    c = f(x_estrela)/exponencial(x_estrela)


    u_exp = np.random.uniform()
    y = -np.log(1-u_exp)
    u = np.random.uniform()

    while c*u*y > f(y):
        u_exp = np.random.uniform()
        y = -np.log(1-u_exp)
        u = np.random.uniform()

    valor = y
    moeda = np.random.uniform()
    if moeda>.5:
        return valor
    else:
        return -valor
```

```python
def Normal(mi,sigma2):
    u1 = np.random.uniform()
    u2 = np.random.uniform()

    u_2pi = 2*np.pi*u1
    t = -np.log(u2)
    x = np.sqrt(2*t)*np.cos(u_2pi)

    return np.sqrt(sigma2)*x + mi
```

```python
#Beta
def Beta(alpha,beta):
    def f(x,alpha,beta):
        B = math.gamma(alpha+beta)/(math.gamma(alpha)*math.gamma(beta))
        f = (x**(alpha-1) * (1-x)**(beta-1))/B
        return f

    y_star = (alpha - 1)/(alpha+beta-2)

    c = f(y_star,alpha,beta)

    y = np.random.uniform()
    u = np.random.uniform()

    while u >= f(y,alpha,beta)/c:
        y = np.random.uniform()
        u = np.random.uniform()

    x = y
    return x
```
