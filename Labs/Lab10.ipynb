## Exercícios — Simulação de Monte Carlo

Em todos os exercícios, utilize $n=10^4$ simulações independentes, salvo indicação contrária.

### Exercício 1 — Estimação de integrais

Considere a integral

$$
I=\int_0^1 e^{-x^2}\,dx.
$$

Utilizando $X\sim U(0,1)$, estime $I$ por Monte Carlo, calcule a variância do estimador e construa um intervalo de confiança de 95%.

### Exercício 2 — Estimação de probabilidades

Considere $X\sim N(0,1)$ e a probabilidade

$$
p=P(X>1).
$$

Estime $p$ por Monte Carlo, calcule a variância do estimador e construa um intervalo de confiança de 95%.

### Exercício 3 — Estimação de $\pi$

Considere $(X,Y)$ uniformemente distribuído em $[0,1]^2$ e o evento

$$
A=\{X^2+Y^2\leq 1\}.
$$

Utilize Monte Carlo para estimar $\pi$, calcule a variância do estimador e construa um intervalo de confiança de 95%.

### Exercício 4 — Estimação de média e variância

Considere $X\sim U(0,1)$.

Estime $E[X]$ e $\operatorname{Var}(X)$ por Monte Carlo. Calcule a variância do estimador da média e construa um intervalo de confiança de 95% para $E[X]$.

### Exercício 5 — Integração em duas dimensões

Considere a integral

$$
I=\int_0^1\int_0^1 e^{-(x^2+y^2)}\,dx\,dy.
$$

Utilizando $(X,Y)\sim U([0,1]^2)$, estime $I$ por Monte Carlo, calcule a variância do estimador e construa um intervalo de confiança de 95%.

### Exercício 6 — Esperança de uma função

Considere $X\sim \operatorname{Exp}(1)$ e

$$
\mu=E[e^{-X}].
$$

Estime $\mu$ por Monte Carlo, calcule a variância do estimador e construa um intervalo de confiança de 95%. Compare com o valor teórico.

### Exercício 7 — Integral imprópria

Considere a integral

$$
I=\int_0^\infty \frac{e^{-x}}{1+x}\,dx.
$$

Utilizando $X\sim \operatorname{Exp}(1)$, estime $I$ por Monte Carlo, calcule a variância do estimador e construa um intervalo de confiança de 95%.

### Exercício 8 — Integral oscilatória

Considere a integral truncada

$$
I_L=\int_0^L \frac{\sin(x)}{x}\,dx,
\qquad L=20.
$$

Utilizando $X\sim U(0,L)$, estime $I_L$ por Monte Carlo, calcule a variância do estimador e construa um intervalo de confiança de 95%. Compare com $\pi/2$.


```python
import numpy as np
import matplotlib.pyplot as plt
```

EXERCÍCIO 1


```python
#Amostra do Exercício 1
N = 10**4
amostra = []
for i in range(N):
  x = np.random.uniform()
  amostra.append(x)

amostra_f = []
for x in amostra:
  amostra_f.append(np.exp(-x**2))
```

```python
media = np.mean(amostra_f)
variance = np.var(amostra_f)
standart_deviation = np.sqrt(variance)

IC_min, IC_max = media-2*standart_deviation/np.sqrt(N),media+2*standart_deviation/np.sqrt(N)
```

```python
print(f"Estimativa Pontual: {media:.4f}")
print(f"Intervalo de Confiaça 95%: {float(IC_min),float(IC_max)}")
```

EXERCÍCIO 2

```python
def g(x):
  return np.exp(-x)


def f(x):
  return (2*(np.exp((-x**2)/2))/np.sqrt(2*np.pi))

c = np.sqrt(2/np.pi)*np.e**(1/2)
```

```python
#Amostra do Exercício 2
amostra_normal = []
N = 10**4
for i in range(N):
  uni = np.random.uniform()
  u = np.random.uniform()
  y = -np.log(1-u)
  while uni*c*g(y) > f(y):
    uni = np.random.uniform()
    u = np.random.uniform()
    y = -np.log(1-u)
  sorteio = np.random.uniform()
  if sorteio > 0.5 :
    X = y
  else:
    X = -y
  amostra_normal.append(X)
```

```python
amostra_p = []
for x in amostra_normal:
  if x>1:
    amostra_p.append(1)
p = np.sum(amostra_p)/N
```

```python
#Intervalo de Confiança
variancia = p*(1-p)/N
IC_lower,IC_upper = p-1.96*np.sqrt(variancia),p+1.96*np.sqrt(variancia)
```

```python
print(f"Estimativa Pontual: {p}")
print(f"Estimativa Intervalar 95%: {float(IC_lower),float(IC_upper)}")
```

Exercício 3

```python
N = 10**4
amostra_circulo = []
for i in range(N):
  x = np.random.uniform()
  y = np.random.uniform()
  if (x**2)+(y**2) <= 1:
    amostra_circulo.append(1)
  else:
    amostra_circulo.append(0)
pi = 4*np.mean(amostra_circulo)
```

```python
var = np.var(amostra_circulo)
var = var/np.sqrt(N)
ic_min, ic_max = pi-1.96*np.sqrt(var), pi+1.96*np.sqrt(var)
print(f"Estimativa Pontual: {pi}")
print(f"Estimativa Intervalar 95%: {float(ic_min),float(ic_max)}")
```

Exercício 7

```python
N = 10**4
amostra_integral = []
for i in range(N):
  unif = np.random.uniform()
  expo = -np.log(1-unif)
  m = 1/(1+expo)
  amostra_integral.append(m)

integral = np.mean(amostra_integral)
var_integral = np.var(amostra_integral)
```

```python
ic_1, ic_2 = integral-1.96*np.sqrt(var_integral/N), integral+1.96*np.sqrt(var_integral/N)
print(f"Estimativa Pontual: {integral}")
print(f"Estimativa Intervalar 95%: {float(ic_1),float(ic_2)}")
```
